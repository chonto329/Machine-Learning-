# Transformaciones

### 9. Transformación 1: Traducir y estandarizar nombres de columnas
Se renombran las 13 columnas del DataFrame de inglés a español manteniendo el formato `snake_case` para mejorar la legibilidad del análisis.

```python
# Diccionario de traducción de columnas
columnas_espanol = {
    'age': 'edad',
    'has_anaemia': 'tiene_anemia',
    'creatinine_phosphokinase_concentration_in_blood': 'cpk_sangre',
    'has_diabetes': 'tiene_diabetes',
    'heart_ejection_fraction': 'fraccion_eyeccion',
    'has_high_blood_pressure': 'tiene_hipertension',
    'platelets_concentration_in_blood': 'concentracion_plaquetas',
    'serum_creatinine_concentration_in_blood': 'creatinina_serica',
    'serum_sodium_concentration_in_blood': 'sodio_serico',
    'is_male': 'es_hombre',
    'is_smoker': 'es_fumador',
    'days_in_study': 'dias_seguimiento',
    'is_dead': 'fallecido'
}

# Aplicar el renombrado
df = df.rename(columns=columnas_espanol)

# Mostrar las primeras filas con los nuevos nombres
df.head()
```

### 10. Transformación 2: Conversión de días a años de seguimiento
Se crea una nueva variable calculada dividiendo los días de seguimiento entre 365 para expresar el tiempo en años.

```python
# Operación matemática directa sobre la columna
df['anos_seguimiento'] = (df['dias_seguimiento'] / 365).round(2)

# Mostrar resultado
df[['dias_seguimiento', 'anos_seguimiento']].head()
```

### 11. Transformación 3: Mapeo de valores a texto explícito
Se transforman las columnas de género y estado vital a texto claro (`Masculino/Femenino` y `Fallecido/Sobreviviente`) para facilitar la interpretación del dataset.

```python
# Crear variables categóricas descriptivas usando .loc
df.loc[:, 'genero'] = df['es_hombre'].map({True: 'Masculino', False: 'Femenino'})
df.loc[:, 'estado_vital'] = df['fallecido'].map({1: 'Fallecido', 0: 'Sobreviviente'})

# Mostrar la comparación de las columnas transformadas
df[['es_hombre', 'genero', 'fallecido', 'estado_vital']].head()
```

### 12. Transformación 4: Agrupamiento y resumen estadístico por grupo
Se agrupan los datos según el `estado_vital` del paciente para calcular los promedios de los principales biomarcadores clínicos.

```python
# Agrupamiento por estado vital y cálculo de promedios
df_resumen = df.groupby('estado_vital').agg({
    'edad': 'mean',
    'fraccion_eyeccion': 'mean',
    'creatinina_serica': 'mean',
    'concentracion_plaquetas': 'mean'
}).rename(columns={
    'edad': 'Edad Promedio',
    'fraccion_eyeccion': 'Fracción Eyección (%)',
    'creatinina_serica': 'Creatinina Promedio',
    'concentracion_plaquetas': 'Plaquetas Promedio'
}).reset_index()

df_resumen
```
