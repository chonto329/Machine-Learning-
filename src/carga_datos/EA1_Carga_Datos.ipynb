# Transformaciones

### 10. Transformación 1: Traducir y estandarizar nombres de columnas
Se renombran las 18 columnas del DataFrame de inglés a español manteniendo el formato `snake_case` para mejorar la legibilidad del análisis.


```python
# Diccionario de traducción de columnas para el dataset de videojuegos
columnas_espanol = {
    'Name': 'nombre',
    'Platform': 'plataforma',
    'Year_of_Release': 'anio_lanzamiento',
    'Genre': 'genero',
    'Publisher': 'editor',
    'NA_Sales': 'ventas_na',
    'EU_Sales': 'ventas_eu',
    'JP_Sales': 'ventas_jp',
    'Other_Sales': 'ventas_otros',
    'Global_Sales': 'ventas_globales',
    'Critic_Score': 'puntuacion_critica',
    'Critic_Count': 'cantidad_criticas',
    'User_Score': 'puntuacion_usuario',
    'User_Count': 'cantidad_usuarios',
    'Developer': 'desarrollador',
    'Rating': 'clasificacion',
    'Year_missing': 'anio_faltante',
    'Is_Hit': 'es_exito'
}

# Aplicar el renombrado
df2 = df2.rename(columns=columnas_espanol)

# Mostrar las primeras filas con los nuevos nombres
df2.head()
```

### 11 Transformación 2: Cálculo de la antigüedad del juego en años
Se crea una nueva variable calculada restando el año de lanzamiento al año actual para expresar la antigüedad de cada título en años.  

```python
# Definir año de referencia / actual
anio_actual = 2026

# Operación matemática directa sobre la columna
df2['antiguedad_anios'] = anio_actual - df2['anio_lanzamiento']

# Mostrar resultado
df2[['nombre', 'anio_lanzamiento', 'antiguedad_anios']].head()
```

### 12. Transformación 3: Porcentaje de ventas regionales sobre el total global
Se calcula la proporción porcentual de las ventas de Norteamérica y Japón en relación con las ventas globales del título.

```python
# Calcular porcentaje de ventas por región respecto al global
df2['pct_ventas_na'] = ((df2['ventas_na'] / df2['ventas_globales']) * 100).round(2)
df2['pct_ventas_jp'] = ((df2['ventas_jp'] / df2['ventas_globales']) * 100).round(2)

# Mostrar resultado
df2[['nombre', 'ventas_globales', 'pct_ventas_na', 'pct_ventas_jp']].head()
```

### 13. Transformación 4: Agrupamiento por Género de Videojuego
Se agrupan los datos según el género para calcular el promedio de ventas globales, el promedio de puntuación de la crítica, el promedio de puntuación de los usuarios y la cantidad total de juegos por género.

```python
# Agrupamiento por género y cálculo de promedios/métricas
df_resumen = df2.groupby('genero').agg({
    'ventas_globales': 'mean',
    'puntuacion_critica': 'mean',
    'puntuacion_usuario': 'mean',
    'nombre': 'count'
}).rename(columns={
    'ventas_globales': 'Ventas Globales Promedio',
    'puntuacion_critica': 'Puntuación Crítica Promedio',
    'puntuacion_usuario': 'Puntuación Usuario Promedio',
    'nombre': 'Total de Videojuegos'
}).reset_index()

df_resumen
```
